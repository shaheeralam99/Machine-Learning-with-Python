```python
import pandas as pd

# ---------------------------
# Create raw dataset
# ---------------------------
data = {
    "student_id": [1,2,3,4,5,6,7],
    "math_score": [78, 85, None, 90, 72, 88, None],
    "english_score": [80, 82, 75, None, 70, 85, 78],
    "study_hours": [5, 7, 4, 8, None, 6, 5],
    "attendance": [90, 95, 85, 98, 80, None, 88]
}

df = pd.DataFrame(data)

# Save raw data
df.to_csv("student_data.csv", index=False)

# ---------------------------
# Data Cleaning
# ---------------------------
df = pd.read_csv("student_data.csv")
df = df.fillna(df.mean(numeric_only=True))

# Feature engineering
df["average_score"] = (df["math_score"] + df["english_score"]) / 2
df["performance_level"] = df["average_score"].apply(
    lambda x: "High" if x >= 80 else "Medium" if x >= 70 else "Low"
)

df

```
